```python
u = [1, 2, 3]
v = [4, 5, 6]
```

```python
def v_add(u, v):
    if(len(u) != len(v)):
        return "diff length"
    n = len(u)
    w = []
    for i in range(0, n):
        val = u[i] + v[i]
        w.append(val)
    return w
print(v_add(u, v))
```

```python
def v_subtract(u, v):
    if(len(u) != len(v)):
        return "diff length"
    n = len(u)
    w = []
    for i in range(0, n):
        val = u[i] - v[i]
        w.append(val)
    return w
print(v_subtract(v, u))
```

```python
def scal_v_mul(a, v):
    n = len(v)
    w = []
    for i in range(0, n):
        val = a * v[i]
        w.append(val)
    return w
a = 3
print(scal_v_mul(a, u))
```

```python
def v_mul(u, v):
    n = len(u)
    w = []
    for i in range (0, n):
        val = u[i] * v[i]
        w.append(val)
    return w
print(v_mul(u, v))
```

```python
# matrix multiplication
A = [[1, 3], [5, 6], [7, 8]]
B = [[1, 4], [4, -1], [2, 5]]
def matrix_mul(A, B):
    n = len(A)
    p = len(A[0])

    ans = []

    for i in range(0, n):
        row = []
        for j in range(0, p):
            val = A[i][j] * B[i][j]
            row.append(val)
        ans.append(row)
    return ans
print(matrix_mul(A, B))
```

```python
# Matrix scalar multiplication
def scalar_mul(A, a):
    n = len(A)
    p = len(A[0])
    ans = []
    for i in range(0, n):
        row = []
        for j in range(0, p):
            val = A[i][j] * a
            row.append(val)
        ans.append(row)
    return ans

print(scalar_mul(A, 4))
print(A)
```

```python
# matrix element multiplication
def elemul(A, B):
    n = len(A)
    p = len(B[0])
    ans = []
    for i in range(0, n):
        row = []
        for j in range(0, p):
            val = A[i][j] * B[i][j]
            row.append(val)
        ans.append(row)
    return ans
print(elemul(A, B))
print(A)
print(B)
```

```python
#matrix multiplication
def matmul(A, B):
    n = len(A)
    a = len(A[0])
    b = len(B[0])
    ans = []
    for i in range(0, n):
        row = []
        for k in range(0, b):
            val = 0
            for j in range(0, a):
                val = val + A[i][j] * B[j][k]
            row.append(val)
        ans.append(row)
    return ans
A = [[1, 2, 3], [4, 5, 6]]
B = [[3, 4], [4, 5], [1, 5]]
print(matmul(A, B))
```

```python
import numpy as np
```

```python
#matrix addition
A = np.array([[1, 2], [2, 3], [4, 5]])
B = np.array([[4, 1], [3, 6], [8, 2]])
C = A + B
print(C)
```

```python
#matrix subtraction
C = A - B
```

```python
#matrix scalar multiplication
a = 3
C = a * A
print(C)
```

```python
#matrix element multiplication
C = np.multiply(A, B)
print(C)
```

```python
#matrix multiplication
A = np.array([[1, 2], [2, 3], [5, 6]])
B = np.array([[1, 2, 3], [4, 5, 6]])
C = np.matmul(A, B)
print(C)
```
